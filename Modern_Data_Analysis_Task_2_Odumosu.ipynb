# Task 2.31 - Array Attributes and Operations

### 11. Perform arithmetic operations on arrays

### Method: For arrays with the same shape, NumPy carries out addition, subtraction, multiplication, and division element by element.


```python
A11 = np.array([
    [2, 4, 6],
    [1, 3, 5]
])

B11 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("A + B:")
print(A11 + B11)

print("\nA - B:")
print(A11 - B11)

print("\nA * B:")
print(A11 * B11)

print("\nA / B:")
print(A11 / B11)
```

### Mathematical interpretation

For corresponding elements $a_{ij}$ and $b_{ij}$:

\[
(A+B)_{ij}=a_{ij}+b_{ij}
\]

\[
(A-B)_{ij}=a_{ij}-b_{ij}
\]

\[
(A\times B)_{ij}=a_{ij}b_{ij}.
\]

Here, `A * B` is **element-wise multiplication**, not matrix multiplication. Matrix multiplication is demonstrated separately in Task 19.


# Task 2.32 - Distance, Norm, Dot Product

### norm.png

I computed the absolute value of every element in the vector $a$ and selected the largest value, the infinity norm is:

\[
\|a\|_{\infty}=\max_i |a_i|.
\]



```python
a17 = np.array([3, -4, 5, -2])

infinity_norm = np.linalg.norm(a17, ord=np.inf)

print("Vector:")
print(a17)
print("Infinity norm:", infinity_norm)
```

### Verification

The absolute values are:

\[
|a|=[3,4,5,2].
\]

Therefore,

\[
\|a\|_{\infty}=\max(3,4,5,2)=5.
\]

NumPy computes this directly with `np.linalg.norm(..., ord=np.inf)`.


# Task 2.33 - Matrix Operations

### 19. Arithmetic and matrix product (`np.dot`)

### Method: To calculate the product of two compatible matrices $A$ and $B$, I multiplied each element in a row of matrix $A$ and each column of matrix $B$, and then added the resulting products to obtain each element $A$$B$. The matrix product is:
Screenshot 2026-09-25 113839.png



```python
A19 = np.array([
    [1, 2],
    [3, 4]
])

B19 = np.array([
    [5, 6],
    [7, 8]
])

print("A + B:")
print(A19 + B19)

print("\nA * B (element-wise multiplication):")
print(A19 * B19)

print("\nnp.dot(A, B) (matrix product):")
print(np.dot(A19, B19))
```

### Interpretation

The key distinction is:

- `A * B` performs element-wise multiplication.
- `np.dot(A, B)` performs the matrix product.

For example,

\[
(AB)_{11}=(1)(5)+(2)(7)=19,
\]

and

\[
(AB)_{12}=(1)(6)+(2)(8)=22.
\]

This distinction is fundamental in numerical linear algebra.


# Task 2.34 - Aggregation

### 27. Mean, variance, and median

### Methods: I calculated the mean by taking the average of all values, I measured the variance by calculating the distance between each value, the mean and the median was identified by choosing the middle value after arranging the dataset in order.


```python
data27 = np.array([2, -3, 5, 7, -1, 4, 10, -6])

mean_value = np.mean(data27)
variance_value = np.var(data27)
median_value = np.median(data27)

print("Data:")
print(data27)
print("Mean:", mean_value)
print("Variance:", variance_value)
print("Median:", median_value)
```

### Mathematical interpretation

The mean is:

\[
\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i=2.25.
\]

The variance returned by `np.var()` uses the population convention (`ddof=0`):

\[
\sigma^2=\frac{1}{n}\sum_{i=1}^{n}(x_i-\bar{x})^2=24.9375.
\]

After sorting the data,

\[
[-6,-3,-1,2,4,5,7,10],
\]

the median is the average of the two central values:

\[
\frac{2+4}{2}=3.
\]

Thus the median is 3.0.


# Task 28 — Count Elements in `[-5, 5]`

### Objective

Count how many elements of the dataset lie in the inclusive interval

\[
[-5,5].
\]

NumPy Boolean conditions are combined with `&`, and `np.count_nonzero()` counts the elements satisfying the condition.


```python
mask28 = (data27 >= -5) & (data27 <= 5)

selected_values = data27[mask28]
count28 = np.count_nonzero(mask28)

print("Boolean mask:")
print(mask28)
print("Selected values:")
print(selected_values)
print("Count:", count28)
```

### Interpretation

The values satisfying

\[
-5\le x\le5
\]

are:

\[
[2,-3,5,-1,4].
\]

Therefore, the number of elements in the interval is:

\[
\boxed{5}.
\]

This demonstrates Boolean masking, a central NumPy technique for selecting data without explicit Python loops.


# Task 33 — Add a Column of Column Means

### Objective

Calculate the mean of each column and append those means as a new column.

For a matrix, `axis=0` calculates one result for each column.


```python
A33 = A19.copy()

column_means = np.mean(A33, axis=0)
A33_with_means = np.column_stack((A33, column_means))

print("Original matrix:")
print(A33)

print("\nColumn means:")
print(column_means)

print("\nMatrix with column means added:")
print(A33_with_means)
```

### Mathematical interpretation

For

\[
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix},
\]

the first column mean is:

\[
\frac{1+3}{2}=2,
\]

and the second column mean is:

\[
\frac{2+4}{2}=3.
\]

The resulting matrix is therefore:

\[
\begin{bmatrix}
1&2&2\\
3&4&3
\end{bmatrix}.
\]

`np.column_stack()` is used to append the column of means without a Python loop.


# Task 38 — Sort All Elements (`axis=None`)

### Objective

Sort every element in the matrix as one collection.

The assignment specifically requires `axis=None`. This tells NumPy to sort all elements rather than sorting each row or column separately.


```python
sorted_elements = np.sort(A33_with_means, axis=None)

print("Matrix:")
print(A33_with_means)

print("\nAll elements sorted:")
print(sorted_elements)
```

### Interpretation

The matrix

\[
\begin{bmatrix}
1&2&2\\
3&4&3
\end{bmatrix}
\]

contains the values:

\[
[1,2,2,3,4,3].
\]

Sorting all elements gives:

\[
[1,2,2,3,3,4].
\]

The use of `axis=None` is important because the entire array is treated as a single collection of elements.


# Task 42 — Text Saving and Loading

### Objective

Save a NumPy array to a text file using `np.savetxt()` and then load it back using `np.loadtxt()`.

This demonstrates a simple data-persistence workflow.


```python
file_name = "task2_array.txt"

np.savetxt(file_name, A33_with_means)

loaded_array = np.loadtxt(file_name)

print("Array saved to:", file_name)
print("\nArray loaded from the text file:")
print(loaded_array)
```

### Interpretation

The workflow is:

\[
\text{NumPy array}
\rightarrow
\texttt{np.savetxt()}
\rightarrow
\text{text file}
\rightarrow
\texttt{np.loadtxt()}
\rightarrow
\text{NumPy array}.
\]

The loaded array reproduces the values that were saved.

This is useful when numerical data needs to be stored in a simple human-readable text format and later reconstructed as a NumPy array.
